```python
import pulp

# Create problem
model = pulp.LpProblem("Supply_Chain_Optimization", pulp.LpMinimize)

# Decision variables
routes = [(w, s) for w in ["W1", "W2"] for s in ["S1", "S2", "S3"]]
x = pulp.LpVariable.dicts("ship", routes, lowBound=0)

# Costs
costs = {
    ("W1","S1"):2, ("W1","S2"):4, ("W1","S3"):5,
    ("W2","S1"):3, ("W2","S2"):1, ("W2","S3"):7
}

# Objective function
model += pulp.lpSum(x[w, s] * costs[(w, s)] for (w, s) in routes)

# Supply constraints
supply = {"W1":100, "W2":150}
for w in supply:
    model += pulp.lpSum(x[w, s] for s in ["S1","S2","S3"]) <= supply[w]

# Demand constraints
demand = {"S1":80, "S2":120, "S3":50}
for s in demand:
    model += pulp.lpSum(x[w, s] for w in ["W1","W2"]) >= demand[s]

# Solve
model.solve()

# Results
print("Status:", pulp.LpStatus[model.status])

for v in model.variables():
    print(v.name, "=", v.varValue)

print("Total Cost =", pulp.value(model.objective))

model += x["W2","S2"] == 0
```
